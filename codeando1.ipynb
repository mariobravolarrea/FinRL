```python
from finvizfinance.quote import finvizfinance

# 1. Inicializar el objeto para Broadcom (AVGO)
stock = finvizfinance('avgo')

# 2. Obtener toda la información y métricas clave en un diccionario
stock_info = stock.ticker_fundament()

# Mostrar la información básica
print(f"Compañía: {stock_info.get('Company')}")
print(f"Sector: {stock_info.get('Sector')}")
print(f"Precio actual: {stock_info.get('Price')}")
print(f"P/E Ratio: {stock_info.get('P/E')}")
```

```python
from finvizfinance.quote import finvizfinance
from IPython.display import Image, display

# 1. Obtener la información de Tesla
stock = finvizfinance('tsla')

# 2. Descargar el gráfico diario
stock.ticker_charts(timeframe='daily', charttype='candle')

# 3. Mostrar la imagen guardada en Colab
display(Image('tsla.jpg'))
```

```python
from finvizfinance.quote import finvizfinance
from IPython.display import Image, display

# 1. Cargar el ticker de CoreWeave
stock = finvizfinance('crwv')

# 2. Obtener métricas fundamentales
fundamentals = stock.ticker_fundament()
print(f"Compañía: {fundamentals.get('Company')}")
print(f"Precio: {fundamentals.get('Price')}")
print(f"P/E: {fundamentals.get('P/E')}")

# 3. Generar y mostrar el gráfico diario en Colab
stock.ticker_charts(timeframe='daily', charttype='candle')
display(Image('crwv.jpg'))
```
