```python
# 1. Instalar la librería para gráficos financieros
!pip install yfinance mplfinance

import yfinance as yf
import mplfinance as mpf
import pandas as pd

# 2. Descargar datos del último año para Futuros de Petróleo WTI (CL=F)
data = yf.download("CL=F", period="1y")

# Normalize column names for mplfinance
expected_ohlcv_map = {
    'open': 'Open',
    'high': 'High',
    'low': 'Low',
    'close': 'Close',
    'adj close': 'Adj Close', # For cases where Adj Close might be present
    'volume': 'Volume'
}

# Handle potential MultiIndex columns and flatten them
if isinstance(data.columns, pd.MultiIndex):
    # Try to extract the most relevant level for OHLCV names
    # Iterate through levels and check if common OHLCV keywords are present
    level_to_use = -1 # Default to the last level
    for i, level_names in enumerate(data.columns.levels):
        if any(keyword.lower() in str(name).lower() for name in level_names for keyword in expected_ohlcv_map.keys()):
            level_to_use = i
            break
    data.columns = data.columns.get_level_values(level_to_use)

# Clean up column names (strip whitespace and standardize casing)
data.columns = [str(col).strip().replace(' ', '').lower() for col in data.columns]
data = data.rename(columns={k: v for k, v in expected_ohlcv_map.items() if k in data.columns})

# Handle 'Adj Close' to ensure 'Close' column is adjusted if available
if 'Adj Close' in data.columns and 'Close' in data.columns and not data['Adj Close'].equals(data['Close']):
    data['Close'] = data['Adj Close']

# Drop 'Adj Close' as mplfinance only needs 'Close'
if 'Adj Close' in data.columns:
    data.drop(columns=['Adj Close'], inplace=True, errors='ignore')

# Ensure standard OHLCV columns for mplfinance are present
ohlcv_standard = ['Open', 'High', 'Low', 'Close', 'Volume']

# Filter the DataFrame to only include these standard OHLCV columns
# This also handles cases where some columns might be missing from yfinance output
data = data[[col for col in ohlcv_standard if col in data.columns]]

# Ensure all selected columns are numeric, forcing to NaN if conversion fails
for col in data.columns: # Iterate over actual columns after selection
    data[col] = pd.to_numeric(data[col], errors='coerce')

# Drop rows with NaN values in OHLCV columns after coercion
data.dropna(subset=ohlcv_standard, inplace=True)

# 3. Graficar en estilo velas japonesas con promedio móvil de 20 y 50 días
mpf.plot(
    data,
    type='candle',
    style='charles',
    title='Gráfico de Velas - Petróleo Crudo WTI (CL=F)',
    ylabel='Precio (USD / Barril)',
    volume=True,
    mav=(20, 50),
    figratio=(12, 6)
)
```

```python
import yfinance as yf
import mplfinance as mpf
import pandas as pd

# Download WTI Crude Oil futures data
data = yf.download("CL=F", period="1y")

# --- Preprocessing for mplfinance ---

# 1. Flatten MultiIndex columns
# From the kernel state, columns are MultiIndex like ('Price', 'Open'), ('Volume', 'Volume')
if isinstance(data.columns, pd.MultiIndex):
    # This will result in ['Close', 'High', 'Low', 'Open', 'Volume'] for the example
    data.columns = data.columns.get_level_values(1)

# 2. Standardize column names to be exactly 'Open', 'High', 'Low', 'Close', 'Volume'
# Ensure casing is correct and handle 'Adj Close' if present
new_cols = {}
for col in data.columns:
    if col.lower() == 'open':
        new_cols[col] = 'Open'
    elif col.lower() == 'high':
        new_cols[col] = 'High'
    elif col.lower() == 'low':
        new_cols[col] = 'Low'
    elif col.lower() == 'close':
        new_cols[col] = 'Close'
    elif col.lower() == 'adj close':
        new_cols[col] = 'Adj Close' # Temporarily keep 'Adj Close' for adjustment
    elif col.lower() == 'volume':
        new_cols[col] = 'Volume'

data = data.rename(columns=new_cols)

# Handle 'Adj Close' to ensure 'Close' column is adjusted if available
if 'Adj Close' in data.columns and 'Close' in data.columns and not data['Adj Close'].equals(data['Close']):
    data['Close'] = data['Adj Close']

# Drop 'Adj Close' as mplfinance only needs 'Close'
if 'Adj Close' in data.columns:
    data.drop(columns=['Adj Close'], inplace=True, errors='ignore')

# Ensure the required OHLCV columns are present
ohclv_required = ['Open', 'High', 'Low', 'Close', 'Volume']
# Filter data to only include these columns
data = data[[col for col in ohclv_required if col in data.columns]]

# Convert all OHLCV columns to numeric, coercing errors to NaN
for col in data.columns:
    data[col] = pd.to_numeric(data[col], errors='coerce').astype(float)

# Drop rows that have NaN values in any of the required OHLCV columns
data.dropna(subset=ohlcv_required, inplace=True)

# --- End Preprocessing ---

# Plot using a dark, inverted background style
mpf.plot(
    data,
    type='candle',
    style='nightclouds',  # Inverted dark background style
    title='Candlestick Chart - WTI Crude Oil (CL=F)',
    ylabel='Precio (USD / Barrel)',
    volume=True,
    mav=(20, 50),
    figratio=(12, 6)
)
```

```python
import yfinance as yf
import pandas as pd

# Descargar los datos de Nexa Resources (NEXA) para asegurar que 'data' esté poblado
ticker = "NEXA"
data = yf.download(ticker, period="1d", interval="1d")

# Manejar las columnas MultiIndex para asegurar nombres de columna de un solo nivel
# La estructura es típicamente (OHLCV_Name, Ticker_Name), queremos OHLCV_Name (nivel 0)
if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)

# Extraer el precio de cierre y el volumen del último día de negociación
# Después de aplanar, data['Close'] y data['Volume'] devolverán Series correctamente.
# .iloc[-1] obtendrá el valor único, y .item() lo convertirá en un escalar.
last_close = data['Close'].iloc[-1].item()
last_volume = data['Volume'].iloc[-1].item()

last_date = data.index[-1].strftime('%Y-%m-%d')

# Calcular el valor total negociado en USD
traded_usd = last_close * last_volume

print(f"Date: {last_date}")
print(f"Closing Price: ${last_close:.2f}")
print(f"Shares Traded: {int(last_volume):,}")
print(f"Total Traded Value: ${traded_usd:,.2f} USD")
```
